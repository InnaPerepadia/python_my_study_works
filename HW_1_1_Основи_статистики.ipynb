**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
mean_visits = sum(visits) / len(visits)
visits_minus_mean = [(i - mean_visits)**2 for i in visits]
std_visits_simple = (sum(visits_minus_mean) / (len(visits) - 1))**0.5
multiplication = [(sum_purchases[i] - mean_sum_purchases) * (visits[i] - mean_visits) for i in range(len(visits))]
cov = sum(multiplication) / (len(visits) - 1)
r = cov / (std_simple_var1 * std_visits_simple)
print(f" Коефіцієнт кореляції Пірсона за Python: {r}")
```

```python
r_var2 = np.corrcoef(sum_purchases,visits)
print(f" Коефіцієнт кореляції Пірсона за NumPy: {r_var2}")
```

**Коефіцієнт кореляції показує слабкий негативний зв'язок між кількістю відвідувань клієнтів та сумою їх покупок за визначений період. Проте дивлячись на графік розсіювання видно два аномальні значення, де клієнти відвідали магазин невелику кількість разів, а сума їх покупок значно більша за середнє значення. При цьому для інших значень з вибірки не видно чіткого взаємозв'язку між візитами та сумою покупок. Варто прибрати з вибірки ці 2 аномалні значення та зробити перерахунок.**

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```
